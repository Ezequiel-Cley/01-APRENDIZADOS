```python
import pandas as pd
import pyodbc
```

```python
server = r'EZEQUIEL\SQLSERVER'
database = 'Ecommerce'
conexaoDB = pyodbc.connect(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    "Trusted_Connection=yes;"
    )
cursor = conexaoDB.cursor() # Crado curso de comando
```

```python
dados = pd.read_excel(r"C:\Users\Cleydenilson\Documents\Cursos\SSIS\Ordens.xlsx")
```
