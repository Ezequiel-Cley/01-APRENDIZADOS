## Trabalhado com funções

O spark tem diversas funções que facilta a sua transformação, agregrações e manipulação dos dados.

**Funções mais utilizadas:**

    withColumn(): Adiciona uma coluna nova ao seu dataframe

    when(): Cria uma Lógica condicional

    col(): Seleciona uma coluna desejada

    lit(): Cria um valor Literal

    groupby() + agg(): Agrupar dados e agregar calculos 

```python
# Importando Bibliotecas para uso do Spark para usar dentro do VSCODE
import findspark
findspark.init()
from pyspark.sql import SparkSession # Import para inciar o spark
from pyspark.sql.types import * # Todos as configurações do SQL de tipagem no spark
from pyspark.sql.functions import * # Todas as funções de SQL do Spark
# Criação da sessão Spark
spark = SparkSession.builder.appName("PySpark-VSCode-App").getOrCreate()
print("PySpark está pronto para uso!")
```

#### Criando um DataFrame Manual

```python
# Criando de forma manual os dados para tabela (DataFrame)
txt_treino = r'C:\Users\Cleydenilson\Documents\Scripts\01 - APRENDIZADOS\07_PySpark\Data\Arquivo_Treino.txt'
df = spark.read.csv(txt_treino, sep=',', header=True, inferSchema=True)
```

```python
df.show()
```

#### Chamada da Função When()

Criando uma condicional que categoriza meus salarios em alto e baixo com condição se for maior ou igual a 4000 é considerado um valor alto, caso não é baixo

```python
df = df.withColumn('Categoria Salarial', # withColumn (Adiciona uma coluna ao seu DataFrame) 
                   when(col('Salário')>=4000, 'Alto' # Agregando a função col que permite eu apenas chamar qual coluna que quero aplicar a condição
                        ).otherwise('Baixo')) # Otherwise é o resultado caso a condição não seja alcançada
```

```python
df.show()
```

#### Agrupado e agregado dados

Groupby() é responsável por agrupar as informações

agg() é responsável por realizar calculos sendo os mais utilizados
    
    count() # Contar a quantidade de uma coluna
    
    sum() # Soma os dados de uma coluna necessita ser um campo númerico
    
    avg() # Média dos dados de uma coluna necessita ser um campo númerico
    
    max() # Retorna o valor máximo de um número ou o primeiro para valor texto

    min() # Retorna o valor mínimo de um número ou o ultimo para valor texto

```python
df.groupby('Departamento').agg(avg('Salário').alias('Média Salárial')).show()
```
