### Criando uma UDF

UDF = User-Defined Function (Função Definida pelo Usuário), se trata de uma função personalizada com comandos python puro, sem depedencias de outras bibliotecas ou funções spark.

```python
# Importando Bibliotecas para uso do Spark para usar dentro do VSCODE
import findspark
findspark.init()
from pyspark.sql import SparkSession # Import para inciar o spark
from pyspark.sql.types import * # Todos as configurações do SQL de tipagem no spark
from pyspark.sql.functions import * # Todas as funções de SQL do Spark
# Criação da sessão Spark
spark = SparkSession.builder.appName("PySpark-VSCode-App").getOrCreate()
print("PySpark está pronto para uso!")
```

```python
# Realizado a leitura de um arquivo CSV para criar um Dataframe com Spark (DataFrame)
txt_treino = r'C:\Users\Cleydenilson\Documents\Scripts\01 - APRENDIZADOS\07_PySpark\Data\Arquivo_Treino.txt'
df = spark.read.csv(txt_treino, sep=',', header=True, inferSchema=True, multiLine=True)
```

```python
df.toPandas()
```

```python
# 1. Função Python Pura
def aumento_10_porcento(valor):
    if valor is None:
        return None
    return float(valor) * 1.10

#Criando variavel que vai armazenar a UDF
aumento_10_porcento_UDF = udf(aumento_10_porcento, FloatType())  # Comando StringType define o tipo da coluna vai passar a ser após o tratamento.
```

```python
df = df.withColumn("Salário_com_Aumento", aumento_10_porcento_UDF(col("Salário").cast("float")))
```

```python
df = df.withColumn("Salário_com_Aumento", col("Salário") * 1.10)
```
